## Numerical Python: numpy

```python
# Importing numpy
# A package is a collection of modules that aren't always included in the base Python
import numpy as np

# In numpy, an array is a generic term for a multidimensional set of numbers
# np.array() defines 1D array, i.e. a vector

x = np.array([100, 200, 300])
y = np.array([20, 12, 17])

# Elements are added respectively, not concatination
# Vector math
print(x + y)

# Matrices are 2D arrays
# Matrices can also be created through np.matrix()
x = np.array([[1, 2], [3, 4]])
print(x)

# Attributes / associated objects
# Access by var.attribute
# numpy-specific
print(x.ndim)
print(x.dtype) # int64 - x is comprised of 64-bit integers

y = np.array([[1, 2], [3.0, 5.2]])
print(y.dtype) # now float64 - floating point numbers

# Package specific documentation
# np.array?

# Forcing data type
x = np.array([[1, 2], [3, 4]], dtype=float)
print(x)
print(x.dtype)

# Shape
print(x.shape)

# Methods
# Functions associated with an object

# Summing all elements
print(x.sum())

# Tuples can't be modified, lists can be modified
# Reshape the array: pass a tuple with new dimensions (rows, columns)
print('before')
print(x)
print('after')
x_reshape = x.reshape((1, 4))
print(x_reshape)

# Arrays are essentially a sequence of rows, row-major ordering
# The other way around is the column-major ordering

# Access (0-based indexing)
print(x[0, 1])
print(x[1, 1])

# Elements modified in x_reshape are also modified in x
# The two objects occupy the same space in memory
print(x)
print(x_reshape)
x_reshape[0][1] = 20
print(x)
print(x_reshape)

# Modifying a tuple throws an exception
tup = (2, 1, 10)
# tup[2] = 3

# x.shape - array dimensions in a tuple
# x.ndim - number of dimensions
# x.T - transpose
print(x.shape)
print(x.ndim)
print(x.T)

# Functions on arrays
x = np.sqrt(x) # Square root
print(x)
x = x**2 # Square elements
print(x)
print(x**0.5) # Square root, again

# Generating random data
# np.random.normal?
# loc - mean, scale - standard deviation

# Signatures, keywords, positional arguments
# Keywords substitued as in documentation - N(0, 1)
x = np.random.normal(size=50)
print(x)

# N(50, 1)
y = x + np.random.normal(50, 1, 50)
print(y)

# Correlation matrix
# np.corrcoef(x, y)
print(np.corrcoef(x, y))

# Setting a random seed
# rng has all random number generating methods in np.random
print(np.random.normal(scale = 5, size = 2))
print(np.random.normal(scale = 5, size = 2))

rng = np.random.default_rng(1303)
print(rng.normal(scale = 5, size = 2))
rng2 = np.random.default_rng(1303)
print(rng2.normal(scale = 5, size = 2))

# np.mean(), np.var(), np.std()
rng = np.random.default_rng(3)
y = rng.standard_normal(10)

# Mean
print(np.mean(y), y.mean())
# Variartion, divides by the sample size n rather than n - 1
print(np.var(y), y.var(), np.mean((y - y.mean())**2))
# Standard Deviation
print(np.std(y), y.std())

# np.var?

# Works on matrices as well
X = rng.standard_normal((10, 3)) # N(0, 1)
print(X)
print(X.mean(axis=0)) # row-major ordering, column means
print(X.mean(axis=1)) # still row-major ordering, row means
```

## Graphics

```python
from matplotlib.pyplot import subplots

# Plot consists of a figure and axes
# Plot is basically a canvas on which the figures are displayed
# Axes contain information: x-, y-axis, labels, titles etc

# Lineplot
# subplots return a two-tuple: a figure object and relavent axes object
# unpacked the tuple into fig and ax
fig, ax = subplots(figsize=(3, 3))
x = rng.standard_normal(100)
y = rng.standard_normal(100)

ax.plot(x, y)

# Scatterplot
fig, ax = subplots(figsize=(3, 3))
ax.plot(x, y, 'o')

# or
fig, ax = subplots(figsize=(3, 3))
ax.scatter(x, y, marker='o');

# semicolons remove the output text

# Axis name and titles
fig, ax = subplots(figsize=(3, 3))
ax.scatter(x, y, marker='o')
ax.set_xlabel('x-axis label')
ax.set_ylabel('y-axis label')
ax.set_title('title')

# fig contains figure information that can be changed and redisplayed
fig.set_size_inches(4,4)
fig

# plotting multiple plots on the same figure
# sometimes an axis can be shared between the graphs (sharex=True)
fig, axes = subplots(nrows=2, ncols=3, figsize=(15,5))
axes[0,1].plot(x,y,'o')
axes[1,2].scatter(x,y,marker='+')

# subplots?

# saving the figures
# dpi = dots per inch
fig.savefig("Fig.png", dpi=400)
fig.savefig("Fig.pdf", dpi=700)

# Modify x-range
axes[0,1].set_xlim([-1,1])
fig

# Contour plot: plots 3D data
# Basically, a topographical map
# x and y are vectors, z (f here) is a matrix
fig, ax = subplots(figsize=(8,8))

# linspace returns a vector of n numbers ranging from a to b
x = np.linspace(-np.pi, np.pi, 50)
y = x
f = np.multiply.outer(np.cos(y), 1/(1+x**2))

# f is a matrix whose elements correspond to the z value for each (x,y) pairs
ax.contour(x,y,f)

# Increase the resolution by adding more levels
fig, ax = subplots(figsize=(8,8))
ax.contour(x,y,f,levels=45);

# ax.contour?

print(x)
print(y)
print(f)

# Heatmap: color-coded plot whose colors depend on the z-value
fig, ax = subplots(figsize=(8,8))
ax.imshow(f)
```

## Indexing Data

```python
# 2D Matrix
A = np.array(np.arange(16).reshape((4,4)))
print(A)

# [row, column], 0-based indexing
print(A[1,2])

# Indexing Rows, Columns, and Submatrices
print(A[[1,3]]) # 2nd and 4th rows

# : selects all
print(A[:, [0,2]]) # 1st and 3rd columns

# two indexing list are wrong for selecting submatrices
# numpy interprets them as pairs of i,j indices for a series of entries
# A[[1,3],[0,2]] # wrong

# Instead, 'select' a first submatrix and then 'select' another one from it
print(A[[1, 3]][:, [0, 2]]) # 2nd & 4th rows and 1st and 3rd columns

# Convenience function, extracts a submatrix using lists, by creating an
# intermediate mesh object
idx = np.ix_([1,3], [0,2,3])
print(A[idx])

# Much more efficient through slices
# Slicing is not the same as indexing the lists
# Slices can be used to extract objects from arbitrary sequences
# (strings, tuples, lists). Indexing is much more limited
# Different data types at play here; array is not a list
# Slicing
print(A[1:4:2, 0:3:2])

# Boolean Indexing
# Boolean True or False, 1 or 0
# vector of 0s with length of A's first dimension (4)
keep_rows = np.zeros(A.shape[0], bool)
print(keep_rows)

keep_rows[[1,3]] = True
print(keep_rows)

# binary representaion
# np.all checks if all entires are True, np.any checks if any entry is True
np.all(keep_rows == np.array([0,1,0,1]))

# booleans and integers retrievings are treated differently by numpy
# Retrieving the first, second, first, and second rows of A (int approach)
print(A[np.array([0,1,0,1])])
# Retrieving only the second and fourth rows of A (bool approach)
print(A[keep_rows])

# Using mesh
keep_cols = np.zeros(A.shape[1], bool)
keep_cols[[0,2,3]] = True
idx_bool = np.ix_(keep_rows, keep_cols)
print(A[idx_bool])

# Mixing lists ad bool arrays
idx_mixed = np.ix_([1,3], keep_cols)
print(A[idx_mixed])
```
