## Boston Dataset

### A)

```python
from ISLP import load_data
Boston = load_data('Boston')
Boston
```

### B)

```python
print('Rows: {0}. Cols: {1}.'.format(Boston.shape[0], Boston.shape[1]))
# Columns: predictors (crime rate, tax, age, etc)
# Rows: samples (housing values of Boston suburbs)
```

### C)

```python
# pd.plotting.scatter_matrix?
pd.plotting.scatter_matrix(Boston, figsize=(15,15));
```

```python
fig, ax = plt.subplots(ncols = 5, nrows = 5, figsize=(15,15))
Boston.plot(x='indus', y='crim', kind='scatter', ax=ax[0][0])
Boston.plot(x='nox', y='crim', kind='scatter', ax=ax[0][1])
Boston.plot(x='rm', y='crim', kind='scatter', ax=ax[0][2])
Boston.plot(x='age', y='crim', kind='scatter', ax=ax[0][3])
Boston.plot(x='dis', y='crim', kind='scatter', ax=ax[0][4])
Boston.plot(x='rad', y='crim', kind='scatter', ax=ax[1][0])
Boston.plot(x='tax', y='crim', kind='scatter', ax=ax[1][1])
Boston.plot(x='ptratio', y='crim', kind='scatter', ax=ax[1][2])
Boston.plot(x='lstat', y='crim', kind='scatter', ax=ax[1][3])
Boston.plot(x='medv', y='crim', kind='scatter', ax=ax[1][4])

Boston.plot(y='nox', x='age', kind='scatter', ax=ax[2][0])
Boston.plot(y='nox', x='dis', kind='scatter', ax=ax[2][1])
Boston.plot(y='nox', x='lstat', kind='scatter', ax=ax[2][2])
Boston.plot(y='nox', x='medv', kind='scatter', ax=ax[2][3])

Boston.plot(y='rm', x='age', kind='scatter', ax=ax[3][0])
Boston.plot(y='rm', x='dis', kind='scatter', ax=ax[3][1])
Boston.plot(y='rm', x='lstat', kind='scatter', ax=ax[3][2])
Boston.plot(y='rm', x='medv', kind='scatter', ax=ax[3][3])

Boston.plot(y='lstat', x='age', kind='scatter', ax=ax[4][0])
Boston.plot(y='medv', x='age', kind='scatter', ax=ax[4][1])
Boston.plot(y='lstat', x='dis', kind='scatter', ax=ax[4][2])
Boston.plot(y='medv', x='dis', kind='scatter', ax=ax[4][3])

Boston.plot(y='dis', x='age', kind='scatter', ax=ax[3][4])
Boston.plot(y='lstat', x='medv', kind='scatter', ax=ax[4][4])

# Strongest correlations include, larger distance to employment centers -> less
# NO concentration (nox), more rooms per dwelling (rm) -> higher home value (medv)
# higher lower status population -> lower home value (medv)
# Strongest predictor correlations are with crime (next answer)
```

### D)

```python
# Crime pc strongly correlates with age, distance to employment centers (dis)
# accessibility to radial highways (rad), tax, lower status population (lstat),
# and home value (medv). Also, okay predictors for crime are non-retail businesses
# (indus), NO concentration (nox), number of rooms/dwelling (rm), and pupil-teacher
# ratio (ptratio), but these must be used more carefully due to outliars
```

### E)

```python
fig, ax = plt.subplots(ncols=3, figsize=(15,5))
Boston.crim.plot(kind='box', ax=ax[0])
Boston.tax.plot(kind='box', ax=ax[1])
Boston.ptratio.plot(kind='box', ax=ax[2])

# A bunch of Boston neighbourhoods have a significant crime rate, marked as outliars
# below, 8 of which have crime rate > 30. The tax rates are distributed more equaly
# with no specific outliars, but the median to upper quartile range is heavily stretched,
# so a large amount of people (25%) pays twice as much as the lower 50% of people.
# The pupil teacher rates is much more symmetric and the median to min/max values
# range is reasonable, only two outliars with a value of 13.
```
