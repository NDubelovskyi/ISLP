### F)

```python
# Binning
college['Elite'] = pd.cut(college['Top10perc'],
                          [0, 50, 100],
                          labels=['No', 'Yes'])

# pd.cut?
college.Elite
```

```python
# pd.value_counts?
college.Elite.value_counts()
```

```python
# college.boxplot?
college.boxplot(column='Outstate', by='Elite');
```
